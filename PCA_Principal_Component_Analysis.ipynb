Principal Component Analysis -> A technique used to eliminate curse of Dimensionality

```python
import pandas as pd
df = pd.read_csv(r"https://raw.githubusercontent.com/sindhura-nk/Datasets/refs/heads/main/Cars93.csv",na_values=['NA',''],keep_default_na=False)
df.head()
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df.isna().sum()
```

## Separate X and Y features

```python
X = df.drop(columns=['id','Weight'])
Y = df[['Weight']]
```

```python
X.head()
```

```python
Y.head()
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest = train_test_split(X,Y,random_state=21,train_size=0.85)
```

## Data preprocessing and Data cleaning

```python
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder

cat = list(X.select_dtypes(include='object'))
con = list(X.select_dtypes(include='number'))

num_pipe = make_pipeline(SimpleImputer(strategy='median'),StandardScaler())

cat_pipe = make_pipeline(SimpleImputer(strategy='most_frequent'),OneHotEncoder(handle_unknown='ignore',sparse_output=False))

pre = ColumnTransformer([
    ('cat',cat_pipe,cat),
    ('con',num_pipe,con)
]).set_output(transform='pandas')
pre
```

```python
pre.fit(xtrain)
```

```python
xtrain_pre = pre.transform(xtrain)
xtest_pre = pre.transform(xtest)
```

```python
xtrain_pre.shape[1]
```

## PCA components

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=3)
pca.fit(xtrain_pre)
```

```python
xtrain_pca = pca.transform(xtrain_pre)
xtrain_pca
```

```python
pca.explained_variance_
```

```python
len(pca.feature_names_in_)
```

## Converting  array into dataframe

```python
X_pca = pd.DataFrame(xtrain_pca)
X_pca
```

```python
X_pca.columns = ['PC1','PC2','PC3']
X_pca
```

```python
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(n_estimators=50,
                              max_depth=5)
model.fit(X_pca,ytrain['Weight'])
```

```python
model.score(X_pca,ytrain)
```

```python
Xtest_pca = pca.transform(xtest_pre)
Xtest_pca
```

```python
Xtest_pca = pd.DataFrame(Xtest_pca)
Xtest_pca.columns = ['PC1','PC2','PC3']
```

```python
model.score(Xtest_pca,ytest)
```
